```python
# 1.3.1 QR Decomposition
# QR decomposition is a convinient and useful way to solve
# the linear least squares problem

import numpy as np
from IPython.display import display, Markdown

A = np.array([
    [1,1],
    [1,0],
    [0,1],
    [1,2]
])

Q, R = np.linalg.qr(A) # perform QR decomposition on matrix A and store the two resulting matricies in Q & R

print("\n")
display(Markdown("A ="))
print(A)

print ("\n_______________________________________")
print("\n")

display(Markdown("\nQ ="))
print(Q)

display(Markdown("\nR ="))
print(R)

print ("\n_______________________________________")
print("\n")

display(Markdown("\nQ @ R ="))
print(Q @ R)

#ROUND Q TO INTEGERS SO IT COULD BE MORE EASILYU SEEN THAT QR = A
QmultR = np.round(Q @ R).astype(int)

print ("\n_______________________________________")
print("\n")

display(Markdown("\nQ @ R rounded to integers = "))
print(QmultR)

# CHECK IF Q HAS ORTHONORMAL COLUMNS:
print ("\n_______________________________________")
print("\n")

display(Markdown("\nDoes Q have orthomnormal columns?"))
print("\n",np.allclose(Q.T @ Q, np.eye(Q.shape[1])))

print("\n")

# CHECK IF QR = A
display(Markdown("\nDoes Q @ R Recunstruct A?"))
print("\n",np.allclose(Q @ R, A))






```
