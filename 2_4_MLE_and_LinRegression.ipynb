# 2.4.1 MLE For Random Samples

  MLE for random samples us a method used to find the parameter values that make observed random samples most probable.

  Def_241.png


```python
import numpy as np
import matplotlib.pyplot as plt

print("_______________________________________________________________________________________")
print("\n")
print("2.4.1. MLE for random samples")
print("_______________________________________________________________________________________")
print("\n")

np.set_printoptions(precision=4, suppress=True)

# Random number generator:
rng = np.random.default_rng(100)

# Sample size:
n = 50
# True rate parameter
true_lambda = 0.5

# Generate an Exponential random sample
sample = rng.exponential(
    scale=1 / true_lambda,
    size=n
)

# Calculating the MLE of lambda
lambda_hat = 1 / sample.mean()

# Possible values of lambda for the graph
grid = np.linspace(0.01, 1.5, 200)

# Calculate the log-likelihood
log_likelihood = (
    n * np.log(grid)
    - grid * sample.sum()
)

# Printing Results:
print("Random sample =")
print(sample)
print("\n")
print("_______________________________________________________________________________________")
print("\n")
print("\nSample size =", n)
print("Sample mean =", sample.mean())
print("True lambda =", true_lambda)
print("MLE lambda-hat =", lambda_hat)
print("\n")
print("_______________________________________________________________________________________")
print("\n")

# Plot the log-likelihood(solid lin) & where MLE is (vertical dashed line):
plt.plot(grid, log_likelihood)

plt.axvline(
    lambda_hat,
    color="tab:green",
    linestyle="--",
    label="MLE"
)

plt.xlabel("Lambda")
plt.ylabel("Log Likelihood")
plt.title("Exponential Log-Likelihood for a Random Sample")
plt.legend()
plt.show()

print("\n")
print("_______________________________________________________________________________________")
print("\n")
```

**What is happening in the above code?**

Here, we have an Exponential(λ) random variable.

Xᵢ,...,Xₙ where in this case, n = 50.

We then calculate ̄X = (1/n) ∑ Xᵢ.
Our MLE is: ̂λ = (1 / ̄X).
We rhen graph: ℓ(λ) = nln(λ) - λ ∑ Xᵢ.
Peak of the log - likelihood curve occurs at λ = ̂λ  .
Our extimate is the value of λ that maximizes the likelihood of the observed data.

# 2.4.2 Linear Regression
Given data points {(**x**ᵢ, yᵢ)}ⁿ ᵢ=1, we seek an affine function to fit the data and each **x**ᵢ = (xᵢ1,...,xᵢₚ).  


```python

print("_______________________________________________________________________________________")
print("\n")
print("2.4.2. Linear Regression")
print("_______________________________________________________________________________________")
print("\n")

# x-values:
n_reg = 40

x = np.linspace(0, 10, n_reg)

# True regression parameters:
true_beta_0 = 2.0
true_beta_1 = 1.5

# random errors from a Normal distribution:
errors = rng.normal(
    loc=0,
    scale=2,
    size=n_reg
)

# y-values:
y = true_beta_0 + true_beta_1 * x + errors



x_bar = x.mean()
y_bar = y.mean()

beta_1_hat = np.sum(
    (x - x_bar) * (y - y_bar)
) / np.sum(
    (x - x_bar)**2
)

beta_0_hat = y_bar - beta_1_hat * x_bar

y_hat = beta_0_hat + beta_1_hat * x

residuals = y - y_hat

print("\nLINEAR REGRESSION")

print("\nTrue intercept =", true_beta_0)
print("Estimated intercept =", beta_0_hat)

print("\nTrue slope =", true_beta_1)
print("Estimated slope =", beta_1_hat)

print("\nResiduals =")
print(residuals)


plt.scatter(
    x,
    y,
    label="Random sample"
)

plt.plot(
    x,
    y_hat,
    label="Regression line"
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Simple Linear Regression")
plt.legend()
plt.show()

print("\n")
print("_______________________________________________________________________________________")
```
