*3.3.1 Necessary and Sufficient Conditions of Local Minimizers*


---

Screen Shot 2026-10-19 at 3.14.41 PM.png

Here, we will focus on local minimizers.



```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=5, suppress=True)
```

```python
print("______________________________________________________________________")
print("LOCAL MINIMIZERS")
print("______________________________________________________________________")
print("\n")

def f(x):
    return x**4 - 8*x**2 + 3

def fp(x):
    return 4*x**3 - 16*x

def fpp(x):
    return 12*x**2 - 16

critical_points = np.array([-2, 0, 2])

for c in critical_points:
    print(f"x={c}: f'(x)={fp(c)}, f''(x)={fpp(c)}, f(x)={f(c)}")

print("\n")

x = np.linspace(-3.5, 3.5, 400)

plt.plot(x, f(x))
plt.scatter(critical_points, f(critical_points), color="tab:red")
plt.title("Local minimizers and a local maximizer")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```


The first critical points are found by setting the first derivative of our function $f(x) = x^4-8x^2+3$ to $0$.  

$f'(x) = 4x^3 - 16x = 0$

We get: $x = -2, 0, 2$

We can see, with the second derivative, which values of $x$ are local minimizers and which are local maximizers.

Our function has two local minimizers at: $x = -2$ and $x = 2$
Our function has a local maximizer at $x = 0$

*3.3.2. Convexity and global minimizers*

---


Screen Shot 2026-10-19 at 3.20.37 PM.png


```python
print("______________________________________________________________________")
print("LOCAL MINIMIZERS")
print("______________________________________________________________________")
print("\n")

def g(x):
    return (x + 1)**2 + 3

x = np.linspace(-5, 3, 300)

plt.plot(x, g(x))
plt.axvline(-1, color="tab:red", linestyle="--", label="global minimizer")
plt.title("A convex function with one global minimizer")
plt.xlabel("x")
plt.ylabel("g(x)")
plt.legend()
plt.show()

```

Above, we have the function $g(x) = (x + 1)^2 + 3$

$g'(x) = 2x + 2$

We set the derivative equal to 0:
$2(x + 1) = 0$

We get: $x = -1$

The second derivative is $g"(x) = 2$

$g"(x) = 2 > 0$ so the function is strictly convex and has a unique global minimizer.

*3.3.3. Gradient descent*

---

```python
print("______________________________________________________________________")
print("GRADIENT DESCENT")
print("______________________________________________________________________")
print("\n")

def g(x):
    return (x + 3)**2 + 2

def grad_g(x):
    return 2 * (x + 3)

learning_rate = 0.2
current = 2.0
path = [current]

for _ in range(20):
    current = current - learning_rate * grad_g(current)
    path.append(current)

path = np.array(path)

print("Final x =", path[-1])
print("Final objective =", g(path[-1]))

plt.plot(path, marker="o")
plt.axhline(-3, color="tab:red", linestyle="--", label="minimizer")
plt.xlabel("iteration")
plt.ylabel("x value")
plt.title("Gradient descent iterates")
plt.legend()
plt.show()

```

Above, we have the function $$g(x) = (x + 3)^2 + 2$$

$g'(x) = 2x + 6$

Setting out derivative equal to zero, we have:
$2(x+3) = 0$

So, $x = -3$.

Therefore, we have a global minimum at $x = -3$.

Our gradient descent starts at $x = 2$ and repeatedly moves towards the global minimizer at $x = -3$.

The graph above shows the iterates of the gradient descent approaching the red dashed line at the global minimizer.
