*3.3.1 Necessary and Sufficient Conditions of Local Minimizers*



```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=5, suppress=True)
```

```python
print("______________________________________________________________________")
print("LOCAL MINIMIZERS")
print("______________________________________________________________________")
print("\n")

def f(x):
    return x**4 - 8*x**2 + 3

def fp(x):
    return 4*x**3 - 16*x

def fpp(x):
    return 12*x**2 - 16

critical_points = np.array([-2, 0, 2])

for c in critical_points:
    print(f"x={c}: f'(x)={fp(c)}, f''(x)={fpp(c)}, f(x)={f(c)}")

print("\n")

x = np.linspace(-3.5, 3.5, 400)

plt.plot(x, f(x))
plt.scatter(critical_points, f(critical_points), color="tab:red")
plt.title("Local minimizers and a local maximizer")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```


The first critical points are found by setting the first derivative of our function $f(x) = x^4-8x^2+3$ to $0$.  

$f'(x) = 4x^3 - 16x = 0$

We get: $x = -2, 0, 2$

We can see, with the second derivative, which values of $x$ are local minimizers and which are local maximizers.

Our function has two local minimizers at: $x = -2$ and $x = 2$
Our function has a local maximizer at $x = 0$
