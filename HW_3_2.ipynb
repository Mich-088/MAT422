# 3.2.1 Limits and Continuity

Continuous_Def.png

```python
import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=6, suppress=True)


```

```python
print("_________________________________________________________________________________________")
print("\n")
print("LIMITS & CONTINUITY")
print("\n")
print("_________________________________________________________________________________________")

h = np.array([0.5, 0.1, 0.01, 0.001])

# Funtions:
left_values = (1 - h)**2 + 1
right_values = 2 * (1 + h) + 2

print("h values:", h)
print("\n")
print("Left-hand values:", left_values)
print("\n")
print("Right-hand values:", right_values)
print("\n")

print("Left-hand limit approaches 2.")
print("\n")
print("Right-hand limit approaches 4.")
print("\n")

# Graphing jump discontinuity
x = np.linspace(-2, 4, 400)

y = np.where(x < 1, x**2 + 1, 2*x + 2)

plt.plot(x, y, label="f(x)")

# Open circle for left-hand limit:
plt.scatter(1, 2, facecolors="white",
            edgecolors="blue", s=100, zorder=5)

# Closed circle for right-hand limit:
plt.scatter(1, 4, color="red", s=100,
            zorder=5)

plt.axvline(1, color="black",
            linestyle="--", linewidth=0.8)

plt.title("Jump discontinuity at x = 1")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

In the function above, there is a jump discontinuity at x = 1.
Therefore the function is not continuous.  
As x approaches 1 from the left, the function approaches 3, but it does not include 2.  
As x approaches 1 from the right, the function approaches 4 and the point (1,4) is included.


# 3.2.2 Derivatives

Deriv_Def.png

```python
print("_________________________________________________________________________________________")
print("\n")
print("DERIVATIVES")

def f(x):
    return 2*x**3 - 4*x**2 + 3*x + 1

# Where derivative is evaluated:
point = 2.0

steps = np.array([1e-1, 1e-2, 1e-3, 1e-4])

numeric_derivatives = (
    f(point + steps) - f(point - steps)
) / (2 * steps)

# Exact derivative:
exact_derivative = 6*point**2 - 8*point + 3

# Absolute errors:
errors = np.abs(numeric_derivatives - exact_derivative)
print("_________________________________________________________________________________________")
print("Step sizes:", steps)
print("\n")
print("Numeric derivative estimates:", numeric_derivatives)
print("\n")
print("Exact derivative:", exact_derivative)
print("\n")
print("Errors:", errors)
print("_________________________________________________________________________________________")

```

In the code above, we differentiate $f(x) = 2x^3-4x^2+3x+1$.
To do this, we pick a point to differentiate on.  In this case, 2.0.
We use the central difference formula to approximate the derivatives.
Our step sizes have values of $h$ that get smaller and smaller.
The smaller our step size, the more accurate out derivative.
We can see this in how the absolute errors get closer and closer to zero the smaller the step size is.

# 3.2.3 Taylor's Theorem

Taylor's theorem gives an approximation of a differentiable function around
a given point by a polynomial.



```python

print("_________________________________________________________________________________________")
print("\n")
print("TAYLOR'S THEOREM")
print("\n")
print("_________________________________________________________________________________________")

# x vals:
x = np.linspace(-2, 2, 400)

# Third-degree Taylor polynomial for sin(x)
taylor_sin = x - x**3 / 6

# Plot funct & approx:
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, taylor_sin, "--",
         label="third-degree Taylor polynomial at 0")

plt.xlabel("x")
plt.ylabel("value")
plt.title("Taylor approximation of sin(x)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Calc max error on [-1, 1]
mask = np.abs(x) <= 1
max_error = np.max(
    np.abs(np.sin(x[mask]) - taylor_sin[mask])
)

print("Max error on [-1, 1] =", max_error)


```

Above, a third-degree Taylor polynomial is used to approximate the $sine$ function.
The sine function is shown in blue and the approximation in red to compare the accuracy of the approximation against the $sine$ funtion.
In addition, the maximum error is shown between $[-1,1]$ so that the error can be seen numerically as well.
