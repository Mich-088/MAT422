```python
# 1.3.3 Linear Regression
# Suppose we are given data points  {(xi,yi)} n, i=1 with each xi = (xi1,...,xid)^(T)
# in Linear Regression we seek to find an affine funtion to fit the data
# in other words, we seek a mathimatical function made up of a linear function
# plus a constant offset

import numpy as np
from IPython.display import display, Markdown
import matplotlib.pyplot as plt

# Data: x's & y's
x = np.array([1.3, 2.5, 3.3, 4.7, 5.2], dtype = float)
y = np.array([2.2, 3.3, 3.5, 5.5, 5.7], dtype = float)

# Constructing matrix A
# Fist column: all 1's
# Second column: x - values
A = np.column_stack((np.ones(len(x)), x))

print("=========================================================================")
print("\n")
display(Markdown("                   LINEAR REGRESSION"))


print("___________________________________________________")
print("\n")
display(Markdown("A ="))
print(A)


print("\n")
print("___________________________________________________")
print("\n")
display(Markdown("\ny ="))
print (y)

#Utilizing QR Decomposition
Q, R = np.linalg.qr(A)

print("\n")
print("___________________________________________________")
print("\n")
display(Markdown("\nQ ="))
print(Q)

print("\n")
display(Markdown("\nR ="))
print(R)

print("\n")
print("___________________________________________________")
print("\n")
# Computing Q^(T)
QT_y = Q.T @ y
display(Markdown("\nQ.T @ y ="))
print(QT_y)

print("\n")
print("___________________________________________________")
print("\n")
# Using Q^(T) (QT_y), we will solve Rβ = Q^(T)_y
beta = np.linalg.solve(R, QT_y)
display(Markdown("\nRegression coefficients: "))
print("beta_0 = ", beta[0])
print("beta_1 = ", beta[1])
print("\n")
print("___________________________________________________")
print("\n")

# Predicted vals
y_vector = A @ beta
# Residual vector
residual_vector = y - y_vector

display(Markdown("\nPredicted values y_vector:"))
print(y_vector)
print("\n")
display(Markdown("\nResidual vector y - y_vector:"))
print(residual_vector)
print("\n")
display(Markdown("\nResidual norm:"))
print(np.linalg.norm(residual_vector))
print("\n")
print("___________________________________________________")
print("\n")

# Is the residual orthogonal to columns of Matrix A?
display(Markdown("\nIs the residual orthogonal to columns of Matrix A?"))
print("A.T @ residual:")
print(A.T @ residual_vector)

print("\n")
print("=========================================================================")
print("\n")

#==========================================================================
#                                  PLOT
#==========================================================================

plt.scatter(x,y, label="Data points")

plt.plot(x, y_vector, label="Linear Regression Line")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression using QR Decomposition")
plt.legend()
plt.grid()
plt.show()











```
